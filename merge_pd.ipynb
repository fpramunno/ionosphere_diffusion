```python
import pandas as pd                                                                                                                                                      
from pathlib import Path
from tqdm import tqdm
                                                                                                                                                   
DATA_DIR = Path('/users/framunno/data/ionosphere')                                                                                                                       
OUTPUT_FILE = DATA_DIR / 'combined_f1m_m1m_2020_2025_interpolated.csv'
                                                                                                                                                                        
years = [2020, 2021, 2022, 2023, 2024, 2025]                                                                                                                             
                                                                                                                                                                        
dfs = []                                                                                                                                                                 
for year in years:
    path = DATA_DIR / f'combined_f1m_m1m_{year}_interpolated.csv'
    print(f'Loading {year}...', end=' ')
    df = pd.read_csv(path)                                                                                                                                               
    print(f'{len(df):,} rows')
    dfs.append(df)                                                                                                                                                       
                
combined = pd.concat(dfs, ignore_index=True)                                                                                                                             
combined['time'] = pd.to_datetime(combined['time'])
combined = combined.sort_values('time').reset_index(drop=True)                                                                                                           
                
print(f'\nTotal rows: {len(combined):,}')                                                                                                                                
print(f'Time range: {combined["time"].min()} -> {combined["time"].max()}')
print(f'Saving to {OUTPUT_FILE}...')                                                                                                                                     
                
combined.to_csv(OUTPUT_FILE, index=False)                                                                                                                                
print('Done.')
```

```python
import pandas as pd                                                                                                                                                      
from pathlib import Path
                                                                                                                                                                        
DATA_DIR = Path('/users/framunno/data/ionosphere')                                                                                                                       
OUTPUT_FILE = DATA_DIR / 'DSCOVR_ORBIT_PRE_2020_2025.csv'
                                                                                                                                                                        
years = [2020, 2021, 2022, 2023, 2024, 2025]                                                                                                                             

dfs = []                                                                                                                                                                 
for year in years:
    path = DATA_DIR / f'DSCOVR_ORBIT_PRE_{year}.csv'
    print(f'Loading {year}...', end=' ')
    df = pd.read_csv(path, comment='#')
    df.columns = df.columns.str.strip()                                                                                                                                  
    print(f'{len(df):,} rows')
    dfs.append(df)                                                                                                                                                       
                
combined = pd.concat(dfs, ignore_index=True)                                                                                                                             
combined['EPOCH_yyyy-mm-ddThh:mm:ss.sssZ'] = pd.to_datetime(
    combined['EPOCH_yyyy-mm-ddThh:mm:ss.sssZ']                                                                                                                           
)               
combined = combined.sort_values('EPOCH_yyyy-mm-ddThh:mm:ss.sssZ').reset_index(drop=True)                                                                                 
                                                                                                                                                                        
print(f'\nTotal rows: {len(combined):,}')
print(f'Time range: {combined.iloc[0, 0]} -> {combined.iloc[-1, 0]}')                                                                                                    
print(f'Saving to {OUTPUT_FILE}...')                                                                                                                                     

combined.to_csv(OUTPUT_FILE, index=False)                                                                                                                                
print('Done.')  
```
